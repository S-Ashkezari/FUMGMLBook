## 1. Introduction to PyTorch

PyTorch is an open-source deep learning framework that is popular for its dynamic computation graph and user-friendly API. 

### 🔄 Difference Between PyTorch and TensorFlow

| Feature               | PyTorch                          | TensorFlow                       |
|-----------------------|----------------------------------|----------------------------------|
| **Computation Graph** | Dynamic (define-by-run)          | Static (define-then-run)         |
| **Debugging**         | Easier (Python-native)           | More complex (requires `tf.function`) |
| **Ease of Use**       | Pythonic and intuitive           | Verbose, more control in deployment |
| **Community**         | Strong in academia and research  | Strong in industry and deployment |
| **Integration**       | Works well with Python libraries | Better support for mobile/TPU     |
| **Model Export**      | TorchScript / ONNX               | SavedModel / TFLite              |

> 🔎 **Summary**: PyTorch is usually preferred in research, while TensorFlow shines in production environments. PyTorch Geometric, in particular, has made PyTorch the leading choice in Graph ML research.

### 🧮 PyTorch Tensors vs TensorFlow Tensors

Both libraries use **tensors** as their core data structure, but syntax and behavior differ.

#### PyTorch Example:

```python
import torch

# Creating tensors
a = torch.tensor([[1.0, 2.0], [3.0, 4.0]])
b = torch.ones_like(a)

# Operations
print("Addition:\n", a + b)
print("Elementwise multiplication:\n", a * b)
print("Matrix multiplication:\n", torch.matmul(a, b.T))
```

#### TensorFlow Example:

```python
import tensorflow as tf

# Creating tensors
a = tf.constant([[1.0, 2.0], [3.0, 4.0]])
b = tf.ones_like(a)

# Operations
print("Addition:\n", a + b)
print("Elementwise multiplication:\n", a * b)
print("Matrix multiplication:\n", tf.matmul(a, tf.transpose(b)))
```

### 🔍 Notes on Tensor Behavior

| Operation        | PyTorch                    | TensorFlow                 |
|------------------|----------------------------|----------------------------|
| `.shape`         | `tensor.shape`             | `tensor.shape`             |
| Gradient Tracking| `requires_grad=True`       | `tf.GradientTape` context  |
| Indexing         | Python-like: `tensor[0, 1]`| Similar                    |
| GPU Support      | `.cuda()`                  | Runs on GPU automatically if available |

### 🔧 Tensors and Operations in PyTorch

```python
import torch

# Create tensors
a = torch.tensor([1.0, 2.0, 3.0])
b = torch.tensor([0.5, 1.0, 1.5])

# Basic operations
print("Addition:", a + b)
print("Dot product:", torch.dot(a, b))
print("Matrix multiplication:", torch.matmul(a.view(1, 3), b.view(3, 1)))
```

### 🧠 Neural Networks with `nn.Module`

```python
import torch.nn as nn

class SimpleNN(nn.Module):
    def __init__(self):
        super(SimpleNN, self).__init__()
        self.linear = nn.Linear(3, 2)

    def forward(self, x):
        return torch.relu(self.linear(x))

model = SimpleNN()
x = torch.tensor([[1.0, 2.0, 3.0]])
output = model(x)
print(output)
```

## 🔍 Autograd and Gradients

```python
x = torch.tensor([2.0, 3.0], requires_grad=True)
y = x[0]**2 + 3 * x[1]
y.backward()
print(x.grad)  # Gradient of y with respect to x
```

Use `.backward()` to compute gradients automatically from the computation graph.

## 💾 Saving and Loading Models

```python
# Save model
torch.save(model.state_dict(), "model.pth")

# Load model
model.load_state_dict(torch.load("model.pth"))
model.eval()
```

## 📦 Custom Loss Function

```python
class MyLoss(nn.Module):
    def forward(self, pred, target):
        return torch.mean((pred - target) ** 2)

loss_fn = MyLoss()
pred = torch.tensor([2.0, 3.0])
target = torch.tensor([1.5, 2.5])
print(loss_fn(pred, target))
```
